### 2. Працюємо з елементами груп
Для кожної з груп: $\mathbb{Z}_{10}$, $\mathbb{Z}_{10}^{*}$, $A_4$, $GL_2(\mathbb{Z}_3)$, виконайте наступні кроки:
1. Побудуйте групу.
2. Виберіть довільні елементи $a, b \in G$ та обчисліть добуток $a \cdot b$.
3. Знайдіть обернені елементи $a^{-1}, b^{-1}$ та виконайте перевірку правильності.
4. Побудуйте підгрупу, породжену елементами $a$ та $b$, і визначте її порядок.
5. Знайдіть всі елементи групи, які є оберненими самі до себе, тобто $a^{-1}=a$.
6. Якщо група $G$ не є абелевою, знайдіть такі елементи $a, b \in G$, що $a \cdot b \neq b \cdot a$.

```sage
print("1. Адитивна група Z_10")
G1 = AdditiveAbelianGroup([10])
a1, b1 = G1([2]), G1([3])
print("2. Елементи a, b:", a1, ",", b1)
print("   Добуток a*b (для адитивної групи це сума a+b):", a1 + b1)
print("3. Обернені (протилежні) a^-1, b^-1:", -a1, ",", -b1)
print("   Перевірка a + (-a) == e:", a1 + (-a1) == G1.zero())
print("4. Порядок підгрупи, породженої a та b:", G1.submodule([a1, b1]).order())
print("5. Елементи, обернені самі до себе (a == -a):", [x for x in G1 if x == -x])
is_abelian_G1 = all(x + y == y + x for x in G1 for y in G1)
print("6. Чи є група абелевою?", is_abelian_G1)
if not is_abelian_G1:
    print("   Оскільки група не є абелевою, перевіряємо a+b != b+a:", a1 + b1 != b1 + a1)
```

```sage
print("\n1. Мультиплікативна група Z_10^*")
G2 = Zmod(10).unit_group()
e2 = G2.list()
a2, b2 = e2[1], e2[2] 
print("2. Елементи a, b (в Z_10):", a2.value(), ",", b2.value())
print("   Добуток a*b (в Z_10):", (a2 * b2).value())
print("3. Обернені a^-1, b^-1:", (a2^-1).value(), ",", (b2^-1).value())
print("   Перевірка a * a^-1 == e:", a2 * a2^-1 == G2.one())
print("4. Порядок підгрупи, породженої a та b:", G2.subgroup([a2, b2]).order())
print("5. Елементи, обернені самі до себе:", [x.value() for x in G2 if x^-1 == x])
is_abelian_G2 = all(x * y == y * x for x in G2 for y in G2)
print("6. Чи є група абелевою?", is_abelian_G2)
if not is_abelian_G2:
    print("   Оскільки група не є абелевою, перевіряємо a*b != b*a:", a2 * b2 != b2 * a2)
```

### 3. Підгрупи та їхні властивості

Для кожної з груп $\mathbb{Z}_{24}$, $Q_8$, $D_5$, $S_4$ виконайте наступні кроки:
1. Побудуйте групу.
2. Отримайте та виведіть список усіх підгруп групи.   
3. Визначте кількість підгруп групи з точністю до ізоморфізму.
4. Визначте порядки всіх підгруп групи.
5. Перевірте, чи порядок кожної підгрупи ділить порядок групи.

```sage
print("1. Адитивна група Z_24 (через ізоморфну циклічну групу)")
G1 = CyclicPermutationGroup(24)
subgroups_G1 = G1.subgroups()
print("2. Список усіх підгруп:")
for H in subgroups_G1:
    print("  ", H)

iso_classes_G1 = []
for H in subgroups_G1:
    if not any(H.is_isomorphic(K) for K in iso_classes_G1):
        iso_classes_G1.append(H)
print("3. Кількість підгруп з точністю до ізоморфізму:", len(iso_classes_G1))

orders_G1 = [H.order() for H in subgroups_G1]
print("4. Порядки всіх підгруп:", orders_G1)
print("5. Чи ділить порядок кожної підгрупи порядок групи?", all(G1.order() % o == 0 for o in orders_G1))
```

### 4. Графи Келі

Графом Келі групи $G$ відносно системи твірних $S$ називається орієнтований граф $\Gamma(G,S)$, вершинами якого є елементи групи і кожний $g\in G$ з'єднаний орієнтованим ребро з $gs$ для всіх $s\in S$.

```sage
print("Діедральна група D4")
G1 = DihedralGroup(4)
CG1 = G1.cayley_graph()
print("1. Граф Келі побудовано (вершин:", CG1.order(), ", ребер:", CG1.size(), ")")
v1 = list(CG1)[0]
print("2. Степінь вершин: вхідний =", CG1.in_degree(v1), ", вихідний =", CG1.out_degree(v1))
print("3. Чи є граф вершинно-транзитивним?", CG1.is_vertex_transitive())
print("4. Чи містить гамільтонів цикл?", CG1.is_hamiltonian())
CG1.plot()
```

### 5*. Ймовірність породження симетричної та знакозмінної груп

1. Створіть групу $S_n$ для деякого $n\geq 10$.
2. Проведіть наступний експеримент $m=1000$ разів:<br>
       2.1. Згенеруйте дві випадкові підстановки $a$ та $b$ в $S_n$.<br>
       2.2. Побудуйте групу $H=\langle a,b\rangle$.<br>
       2.3. Перевірте, чи $H$ дорівнює $S_n$, $H=A_n$ або є іншою підгрупою.
3. Обчисліть частоту кожного результату. 
4. Сформулюйте гіпотезу: яка ймовірність породження $S_n$ та $A_n$ двома підстановками.

```sage
print("Ймовірність породження S_n та A_n")
n = 10
m = 1000
G = SymmetricGroup(n)
order_Sn = G.order()
order_An = order_Sn / 2

count_Sn = 0
count_An = 0
count_other = 0

print(f"1. Створено групу S_{n}. Порядок S_n: {order_Sn}, порядок A_n: {order_An}")
print(f"2. Проводимо {m} ітерацій... (це може зайняти кілька секунд)")

for _ in range(m):
    a = G.random_element()
    b = G.random_element()
    H = G.subgroup([a, b])
    H_order = H.order()
    
    if H_order == order_Sn:
        count_Sn += 1
    elif H_order == order_An:
        count_An += 1
    else:
        count_other += 1

freq_Sn = float(count_Sn / m)
freq_An = float(count_An / m)
freq_other = float(count_other / m)

print("\n3. Результати (частоти):")
print(f"   Кількість H = S_n: {count_Sn} (Частота: {freq_Sn:.3f})")
print(f"   Кількість H = A_n: {count_An} (Частота: {freq_An:.3f})")
print(f"   Кількість інших підгруп: {count_other} (Частота: {freq_other:.3f})")

print("\n4. Сформульована гіпотеза (програмний висновок):")

if freq_Sn > 0.7 and freq_An > 0.2 and freq_other < 0.05:
    print("   Аналіз результатів показує:")
    print("   -> Ймовірність того, що дві випадкові підстановки породять групу S_n, наближається до 3/4 (75%).")
    print("   -> Ймовірність породити групу A_n наближається до 1/4 (25%).")
    print("   -> Ймовірність породити іншу підгрупу прямує до нуля.")
else:
    print("   Результати експерименту нестандартні. Гіпотеза: ймовірності розподіляються інакше для даного n.")
```
