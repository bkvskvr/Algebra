### 1. Властивості циклічних груп

Для наступних порядків $n \in \{100, 173, 530, 746, 911, 1024\}$ розгляньте циклічну групу $G = \mathbb{Z}_n$ (або $C_n$).
1. Знайдіть усі дільники $d$ порядку $n$.<br>
2. Для кожного дільника $d$ обчисліть кількість елементів $a \in G$ таких, що $\text{ord}(a) = d$.<br>
3. Порівняйте отриману кількість з теоретичним значенням $\phi(d)$, де $\phi$ --- функція Ойлера.<br>
4. Знайдіть усі підгрупи групи $G$.<br>
5. Обчисліть загальну кількість усіх підгруп.<br>
6. Порівняйте отриману кількість з теоретичним значенням $\tau(n)$, де $\tau$ -- функція кількості дільників числа $n$.

```sage
def cyclic_report(size):
    group = Integers(size)
    distribution = {}
    for element in group:
        order = element.additive_order()
        distribution[order] = distribution.get(order, 0) + 1

    rows = []
    subgroups = []
    for order in divisors(size):
        step = size // order
        members = list(range(0, size, step))
        subgroups.append(members)
        rows.append([order, distribution[order], euler_phi(order),
                     distribution[order] == euler_phi(order)])

    print("\nЦиклічна група Z_%s" % size)
    print("1. Дільники порядку:", divisors(size))
    print("2. Кількість елементів кожного порядку:")
    show(table([row[:2] for row in rows], header_row=["d", "Кількість"]))
    print("3. Порівняння з функцією Ейлера:")
    show(table(rows, header_row=["d", "Кількість", "φ(d)", "Збіг"]))
    print("4. Усі підгрупи та їхні твірні:")
    for members in subgroups:
        generator = 0 if len(members) == 1 else members[1]
        print("<%s> = %s" % (generator, members))
    theoretical = prod(exponent + 1 for prime, exponent in factor(size))
    print("5. Загальна кількість підгруп:", len(subgroups))
    print("6. Порівняння з τ(n):")
    print("τ(%s) = %s; збіг: %s" % (size, theoretical, len(subgroups) == theoretical))

for size in (100, 173, 530, 746, 911, 1024):
    cyclic_report(size)
```

### 2. Порядки елементів в діедральних групах

Для наступних значень $n\in\{5,6,10,12,16,17,20\}$ розгляньте діедральну групу $D_n$.   
1. Знайдіть кількість елементів порядку $d=2$ в групі $D_n$.
2. Для кожного дільника $d>2$ порядку $|D_n|=2n$ знайдіть кількість елементів порядку $d$ у групі $D_n$.<br>

Проаналізуйте результати та дайте відповіді на запитання:
1. Скільки елементів порядку $d=2$ в діедральній групі $D_n$?
2. Скільки елементів порядку $d>2$ в діедральній групі $D_n$?
3. Чи для кожного дільника $d<2n$ порядку групи $D_n$ існує елемент порядку $d$?

```sage
for sides in (5, 6, 10, 12, 16, 17, 20):
    symmetry = DihedralGroup(sides)
    frequency = {}
    for transformation in symmetry:
        order = transformation.order()
        frequency[order] = frequency.get(order, 0) + 1

    rows = []
    for order in divisors(symmetry.order()):
        if order == 1:
            continue
        if order == 2:
            predicted = sides + int(sides % 2 == 0)
        else:
            predicted = euler_phi(order) if sides % order == 0 else 0
        actual = frequency.get(order, 0)
        rows.append([order, actual, predicted, actual == predicted])

    print("\nD_%s, кількість елементів: %s" % (sides, symmetry.order()))
    print("1. Елементи порядку 2:", frequency.get(2, 0))
    print("2. Елементи порядків d > 2:")
    show(table([row for row in rows if row[0] > 2],
               header_row=["d", "Кількість", "За формулою", "Збіг"]))
    absent = []
    for order in divisors(symmetry.order()):
        if order < symmetry.order() and order not in frequency:
            absent.append(order)
    print("Відсутні порядки серед дільників d < 2n:", absent)

print("\nВідповіді на запитання:")
print("1. Порядок 2: n відбиттів і ще одне обертання, якщо n парне.")
print("2. Порядок d > 2: φ(d) елементів при d | n, інакше 0.")
print("3. Не кожен дільник реалізується: у D_6 немає елементів порядку 4.")
```

### 3. Дискретний логарифм

Розглянемо групу $\mathbb{Z}_n^{*}$ для $n=1102158078129785185997827239806$.
1. Знайдіть факторизацію модуля $n$.
2. Перевірте, що група є циклічною, та знайдіть її порядок.
3. Знайдіть найменше натуральне число $a \in \mathbb{N}$ таке, що елемент $\overline{a} \in \mathbb{Z}_n^{*}$ є твірним елементном групи.
4. Перевірте, що порядок елемента $\overline{a}$ дорівнює $\phi(n)$.
5. Виберіть три випадкові елементи $g_1, g_2, g_3 \in \mathbb{Z}_n^{*}$.
6. Для кожного елемента $g \in \{g_1, g_2, g_3\}$ знайдіть дискретний логарифм за модулем $n$, тобто таке число $k$, що $g=\overline{a}^k$ в $\mathbb{Z}_n^{*}$.


```sage
modulus = ZZ(1102158078129785185997827239806)
residues = Integers(modulus)
unit_count = euler_phi(modulus)
cyclic = residues.multiplicative_group_is_cyclic()
print("1. Факторизація n:", factor(modulus))
print("2. Циклічність групи одиниць:", cyclic)
print("Кількість одиниць:", unit_count)
assert cyclic

checks = [unit_count // prime for prime in prime_divisors(unit_count)]
candidate = ZZ(1)
while True:
    root = residues(candidate)
    if root.is_unit() and all(root**power != 1 for power in checks):
        break
    candidate += 1

root_order = root.multiplicative_order()
print("3. Найменший натуральний твірний:", candidate)
print("4. Порядок твірного:", root_order)
print("ord(a) = φ(n):", root_order == unit_count)

samples = []
while len(samples) != 3:
    value = residues(ZZ.random_element(1, modulus))
    if value.is_unit() and value not in samples:
        samples.append(value)

print("5. Три випадкові елементи:")
for index, value in enumerate(samples, 1):
    print("g_%s = %s" % (index, value))

print("6. Дискретні логарифми та перевірка:")
results = []
for value in samples:
    exponent = ZZ(value.log(root)) % unit_count
    recovered = root**exponent
    results.append([value, exponent, recovered, recovered == value])
show(table(results, header_row=["g", "log_a(g)", "a^k", "Перевірка"]))
```

### 4. Одна лінійна група

Розглянемо підгрупу $G$ групи $GL_3(\mathbb{Z})$, породжену двома матрицями:
$$G = \left\langle \begin{pmatrix} 1 & 0 & 1 \\ 0 & 1 & 0 \\ 0 & 0 & 1 \end{pmatrix}, \begin{pmatrix} 1 & 0 & 0 \\ 0 & 1 & 1 \\ 0 & 0 & 1 \end{pmatrix} \right\rangle.$$
Дайте відповідь на наступні питання, обґрунтовуючи відповіді за допомогою обчислень у SageMath:
1. Чи є група $G$ скінченною?
2. Чи є група $G$ абелевою?
3. Які порядки мають елементи групи $G$?
4. Який вигляд мають елементи групи $G$?
5. Чи є група $G$ циклічною?
6. Якій відомій групі ізоморфна група $G$?

```sage
U = matrix(ZZ, 3, 3, [1,0,1, 0,1,0, 0,0,1])
V = matrix(ZZ, 3, 3, [1,0,0, 0,1,1, 0,0,1])
x, y, z, w = var("x y z w")

def element(a, b):
    return matrix(SR, [[1,0,a], [0,1,b], [0,0,1]])

print("Твірні комутують:", U*V == V*U)
print("Загальний елемент M(x,y):")
show(element(x,y))
print("Добуток двох довільних елементів:")
show(element(x,y)*element(z,w))
print("M(x,y)M(z,w) = M(x+z,y+w):",
      element(x,y)*element(z,w) == element(x+z,y+w))
print("M(-x,-y) є оберненим:",
      element(x,y)*element(-x,-y) == identity_matrix(SR,3))

for a, b in [(2,-3), (-1,-2), (0,0)]:
    print("U^(%s)V^(%s) = M(%s,%s):" % (a,b,a,b),
          U**a * V**b == element(a,b))

print("\nВідповіді за пунктами:")
print("1. G нескінченна: U^k = M(k,0), і ці матриці попарно різні.")
print("2. G абелева: множення відповідає додаванню параметрів.")
print("3. M(x,y)^k = M(kx,ky). Порядок одиниці — 1; решти — нескінченний.")
print("4. G = {M(x,y): x,y належать Z}, оскільки M(x,y) = U^x V^y.")
print("5. G не циклічна: якщо M(x,y)^k = U, то kx=1, ky=0, отже y=0.")
print("Тоді жоден степінь M(x,y) не дорівнює V.")
print("6. Ізоморфізм Z × Z → G задається відповідністю (x,y) → M(x,y).")
```

### 5*. Циклічність груп $\mathbb{Z}_n^{*}$

Для кожного $n=2, 3, 4, \ldots, 256$ виконайте наступні кроки, представляючи результати у вигляді таблиці:
1. Знайдіть факторизацію числа $n$.
2. Перевірте, чи є група $\mathbb{Z}_n^{*}$ циклічною.
3. Для циклічних груп $\mathbb{Z}_n^{*}$ знайдіть найменше $a \in \mathbb{N}$ таке, що $\overline{a}$ є твірним групи.
4. Сформулюйте гіпотезу: для яких $n \in \mathbb{N}$ група $\mathbb{Z}_n^{*}$ є циклічною.

```sage
def least_generator(modulus):
    ring = Integers(modulus)
    if not ring.multiplicative_group_is_cyclic():
        return None
    target = euler_phi(modulus)
    for representative in range(1, modulus):
        value = ring(representative)
        if value.is_unit() and value.multiplicative_order() == target:
            return representative

summary = []
for modulus in range(2,257):
    generator = least_generator(modulus)
    summary.append([modulus, factor(modulus),
                    generator is not None,
                    generator if generator is not None else "—"])
print("Пункти 1–3 наведено у відповідних стовпцях таблиці:")
show(table(summary, header_row=["n", "1. Факторизація", "2. Циклічність", "3. Найменший твірний"]))
print("4. Гіпотеза для n ≥ 2: n = 2, 4, p^r або 2p^r, де p — непарне просте, r ≥ 1.")
print("Для n=1 за стандартною домовленістю група одиниць також тривіальна й циклічна.")
print("Таблиця підтверджує гіпотезу до 256; загальне твердження є теоремою.")
```

### 6*. Числа Фібоначчі

Розглянемо підгрупу $H$ групи $GL_2(\mathbb{Z})$, породжену двома матрицями:
$$H = \left\langle \begin{pmatrix} 1 & -1 \\ -1 & 2 \end{pmatrix}, \begin{pmatrix} 3 & 2 \\ 2 & 1 \end{pmatrix} \right\rangle.$$
Дайте відповідь на наступні питання, обґрунтовуючи відповіді за допомогою обчислень у SageMath: 

1. Чи є група $H$ скінченною?
2. Чи є група $H$ абелевою?
3. Які порядки мають елементи групи $H$?
4. Який вигляд мають елементи групи $H$?
5. Чи є група $H$ циклічною? Якщо так, то знайдіть її твірний елемент.

```sage
P = matrix(ZZ, [[1,-1],[-1,2]])
R = matrix(ZZ, [[3,2],[2,1]])
T = P*R
print("Кандидат у твірні T = PR:")
show(T)
print("P = T^(-2):", T**(-2) == P)
print("R = T^3:", T**3 == R)
print("PR = RP:", P*R == R*P)
print("Характеристичний поліном T:", T.charpoly())

def fib_signed(index):
    if index >= 0:
        return fibonacci(index)
    return (-1)**(-index+1)*fibonacci(-index)

rows = []
for exponent in [-4,-2,-1,0,1,2,4]:
    expression = matrix(ZZ, [[fib_signed(exponent+1),fib_signed(exponent)],
                             [fib_signed(exponent),fib_signed(exponent-1)]])
    rows.append([exponent, T**exponent == expression])
show(table(rows, header_row=["k", "Перевірка формули Фібоначчі"]))
print("\nВідповіді за пунктами:")
print("1. H нескінченна: T має власне значення (1+sqrt(5))/2 > 1.")
print("Тому жоден додатний степінь T не дорівнює одиничній матриці.")
print("2. H абелева, оскільки всі її елементи є степенями T.")
print("3. Порядок T^0 — 1; порядок T^k для k ≠ 0 — нескінченний.")
print("4. T^k = [[F_(k+1), F_k], [F_k, F_(k-1)]], k належить Z.")
print("F_0=0, F_1=1, F_(j+1)=F_j+F_(j-1), F_(-j)=(-1)^(j+1)F_j.")
print("Формула випливає з рекурентності при множенні на T та T^(-1).")
print("5. H = <T>: T=PR належить H, а P=T^(-2), R=T^3 належать <T>.")
print("Отже, H — нескінченна циклічна група, ізоморфна (Z,+).")
```
