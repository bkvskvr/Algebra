### 1. Класи суміжності та нормальні підгрупи

Розглянемо групу $S_4$ та її підгрупи:<br>
$\quad$ (a) $H_1=\langle (1,2,3,4)\rangle$;<br>
$\quad$ (b) $H_2=\langle (1,3), (1,2,3,4)\rangle$;<br>
$\quad$ (c) $H_3=K_4$.

Для кожної підгрупи $H$ виконайте наступні кроки:
1. Обчисліть порядок підгрупи $H$ та її індекс у групі $[S_4 : H]$.
2. Знайдіть ліві та праві класи суміжності підгрупи $H$ у $S_4$.
3. Визначте, чи є $H$ нормальною підгрупою $S_4$.
4. Якщо $H$ є нормальною, то побудуйте факторгрупу $G/H$, її таблицю Келі та визначіть структуру.
5. Якщо $H$ не є нормальною, то знайдіть такий елемент $a\in S_4$, що $aH\neq Ha$.


```sage
S4 = SymmetricGroup(4)
r = S4('(1,2,3,4)')
s = S4('(1,3)')

subgroups = {
    'H1': S4.subgroup([r]),
    'H2': S4.subgroup([r, s]),
    'H3': S4.subgroup([S4('(1,2)(3,4)'), S4('(1,3)(2,4)')])
}

def make_cosets(group, subgroup, side):
    remaining = set(group)
    result = []
    while remaining:
        g = min(remaining, key=str)
        part = frozenset(g * h if side == 'left' else h * g for h in subgroup)
        result.append(part)
        remaining.difference_update(part)
    return result

for name, H in subgroups.items():
    left = make_cosets(S4, H, 'left')
    right = make_cosets(S4, H, 'right')
    normal = set(left) == set(right)
    print('\n', name, ': порядок =', H.order(), ', індекс =', S4.order() // H.order())
    print('Ліві класи:')
    for i, coset in enumerate(left, 1):
        print(' L{} = {}'.format(i, sorted(map(str, coset))))
    print('Праві класи:')
    for i, coset in enumerate(right, 1):
        print(' R{} = {}'.format(i, sorted(map(str, coset))))
    print('Нормальна підгрупа:', normal)
    if normal:
        Q = S4.quotient(H)
        representatives = [min(C, key=str) for C in left]
        position = {x: i for i, C in enumerate(left) for x in C}
        print('Факторгрупа має порядок', Q.order())
        print('Таблиця Келі для класів C1, ..., C{}:'.format(len(left)))
        print('    ' + ' '.join('C{:>2}'.format(i) for i in range(1, len(left) + 1)))
        for i, a in enumerate(representatives, 1):
            row = [position[a * b] + 1 for b in representatives]
            print('C{:>2}:'.format(i), ' '.join('C{:>2}'.format(v) for v in row))
        if Q.is_isomorphic(SymmetricGroup(3)):
            print('Структура факторгрупи: S3')
        else:
            print('Структура факторгрупи:', Q)
    else:
        witness = next(g for g in S4 if {g * h for h in H} != {h * g for h in H})
        print('Приклад aH != Ha: a =', witness)
```

### 2. Нормальні підгрупи в діедральних групах

Розгляньте діедральні групи $D_n$ для значень $n\in\{3,4,\dots,30\}$ і виконайте такі кроки:
1. Побудуйте групу $D_n$.
2. Визначте кількість нормальних підгруп у $D_n$.
3. Складіть перелік порядків усіх нормальних підгруп.
4. Визначте алгебраїчну структуру кожної нормальної підгрупи.
5. Знайдіть закономірність і виведіть гіпотезу про кількість та структуру нормальних підгруп в $D_n$.
6. Сформулювати гіпотезу: Скільки нормальних підгруп в $D_n$ та яка їхня структура? Спрогнозувати кількість нормальних підгруп у $D_{470448}$ без виконання обчислень. 

```sage
print(' n | нормальних | порядки | структури')
print('-' * 95)

for n in range(3, 31):
    D = libgap.DihedralGroup(2 * n)
    normal = list(D.NormalSubgroups())
    normal.sort(key=lambda H: (int(H.Order()), str(H)))
    orders = [int(H.Order()) for H in normal]
    kinds = []
    for H in normal:
        size = int(H.Order())
        if bool(H.IsCyclic()):
            kinds.append('C{}'.format(size))
        elif size == 4:
            kinds.append('C2 x C2')
        else:
            kinds.append('D{}'.format(size // 2))
    predicted = len(divisors(n)) + (3 if n % 2 == 0 else 1)
    assert len(normal) == predicted
    print('{:>2} | {:>10} | {} | {}'.format(n, len(normal), orders, kinds))

m = 470448
number = len(divisors(m)) + (3 if m % 2 == 0 else 1)
print('\nДля D_n: tau(n) + 1 при непарному n; tau(n) + 3 при парному n.')
print('470448 =', factor(m))
print('Кількість нормальних підгруп D_470448 =', number)
```

### 3. Гомоморфізм групи матриць у симетричну групу. 

Розглянемо групу $GL_2(\mathbb{Z}_3)$ та множину прямих $M=\{\ell_1, \ell_2, \ell_3, \ell_4\}$ у просторі $\mathbb{Z}_3^2$, що визначаються парами пропорційних ненульових векторів:
\begin{align*}
\ell_1 = \{ (1,0), (2,0) \}, \
\ell_2 = \{ (0,1), (0,2) \}, \
\ell_3 = \{ (1,1), (2,2) \}, \
\ell_4 = \{ (1,2), (2,1) \}.
\end{align*}

1. Побудуйте групу $GL_2(\mathbb{Z}_3)$.
2. Задайте відображення
    $$
    \phi: GL_2(\mathbb{Z}_3) \longrightarrow S_4,
    $$
    яке переводить матрицю у перестановку, що описує її дію на прямі з $M$.
3. Перевірте, що $\phi$ є гомоморфізмом груп.<br>
4. Обчисліть ядро $N=\ker(\phi)$ та образ $\mathrm{Im}(\phi)$.
5. Побудуйте факторгрупу $GL_2(\mathbb{Z}_3)/N$ та порівняйте її з образом $\mathrm{Im}(\phi)$.



```sage
F = GF(3)
G = GL(2, F)
P = SymmetricGroup(4)
lines = [vector(F, v) for v in [(1, 0), (0, 1), (1, 1), (1, 2)]]

def normalized(v):
    c = next(x for x in v if x != 0)
    return tuple(int(x / c) for x in v)

line_ids = {normalized(v): i + 1 for i, v in enumerate(lines)}

def phi(A):
    action = [line_ids[normalized(A.matrix() * v)] for v in lines]
    return P(action).inverse()

checks = all(phi(A * B) == phi(A) * phi(B) for A in G for B in G)
kernel = [A for A in G if phi(A) == P.one()]
image = {phi(A) for A in G}

print('|GL2(Z3)| =', G.order())
print('Відображення є гомоморфізмом:', checks)
print('Ядро:')
for A in kernel:
    print(A.matrix())
print('|ker(phi)| =', len(kernel))
print('|Im(phi)| =', len(image))

z = G(2 * identity_matrix(F, 2))
N = G.subgroup([z])

def matrix_key(A):
    return tuple(int(x) for x in A.matrix().list())

cosets = {}
for A in G:
    key = min(matrix_key(A), matrix_key(z * A))
    cosets.setdefault(key, []).append(A)

coset_map = {key: phi(elements[0]) for key, elements in cosets.items()}
assert all(len(elements) == 2 for elements in cosets.values())
assert all(all(phi(A) == coset_map[key] for A in elements) for key, elements in cosets.items())
assert len(coset_map) == len(set(coset_map.values())) == len(image) == 24
print('|N| =', N.order())
print('|GL2(Z3)/N| =', len(cosets))
print('Факторгрупа ізоморфна образу та S4:', checks and len(image) == P.order())
```

### 5*. Проекції групи $SL_2(\mathbb{Z})$ на $SL_2(\mathbb{Z}_n)$ 

Розглянемо спеціальну лінійну групу $SL_2(\mathbb{Z})$, яка породжується двома матрицями:
$$S = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix},\qquad T = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}. $$

1. Задайте гомоморфізм $\rho_n: SL_2(\mathbb{Z}) \to SL_2(\mathbb{Z}_n)$, який редукує матриці за модулем $n$. 
2. Для значень $n\in\{2, 3, \ldots, 10\}$ перевірте, чи є $\rho_n$ сюр'єктивним, виконавши такі кроки:<br>
    a. Побудуйте групу $G_n = SL_2(\mathbb{Z}_n)$.<br>
    b. Побудуйте групу $H_n = \langle \rho_n(S), \rho_n(T) \rangle$.<br>
    c. Порівняйте порядки груп $H_n$ та $G_n$.



```sage
S = matrix(ZZ, [[0, -1], [1, 0]])
T = matrix(ZZ, [[1, 1], [0, 1]])

def rho(A, n):
    return matrix(Integers(n), 2, 2, A.list())

print(' n | |SL2(Z/nZ)| | |<rho(S),rho(T)>| | сюр’єкція')
print('-' * 60)

for n in range(2, 11):
    Gn = SL(2, Integers(n))
    sn, tn = Gn(rho(S, n)), Gn(rho(T, n))
    Hn = Gn.subgroup([sn, tn])
    same = Hn.order() == Gn.order()
    print('{:>2} | {:>11} | {:>18} | {}'.format(n, Gn.order(), Hn.order(), same))
    assert same

print('Висновок: редукція modulo n сюр’єктивна для всіх перевірених n.')
```
