### 1. Арифметичні операції

а) додайте 25 і 17;<br>
б) помножте 7 на 9;<br>
в) поділіть 144 на 12;<br>
г) знайдіть залишок від ділення 156 на 7;<br>
д) поділіть 156 націло на 7;<br>
е) розкладіть число 2024 на прості множники.

```sage
print("a) 25+17 = ", 25+17)
print("б) 7*9 = ", 7*9)
print("в) 144/12 = ", 144/12)
print("г) залишок від ділення 156 на 7: ", 156%7)
print("д) 156 націло на 7: ", 156//7)
print("е) 2024 на прості множники:", factor(2024)) 
```

###  3. Символічні обчислення та алгебра виразів
а) оголосіть змінні $x=10$ і $y=15$ та обчисліть їхню суму і добуток;<br>
б) оголосіть символічну змінну $z$ і обчисліть вираз $3z^2 + 2z - 5$;<br>
в) спростіть вираз $2x^2 + 4x - 2x^2 + x$;<br>
г) розкрийте дужки у виразі $(x + 2)(x - 3)$;<br>
д) розкладіть на множники тричлен $x^2 - 9x + 20$;<br>
е) розв'яжіть рівняння $x^2 - 4x + 3 = 0$.<br>

```sage
x = 10
y = 15
print("а) сума = ", x+y, "добуток = ", x*y)
var('z')
print("б) вираз = ", 3*z^2 + 2*z - 5)
var('x')
print("в) спрощений вираз = ", simplify(2*x^2 + 4*x - 2*x^2 + x))
print("г) розкриті дужки = ", expand((x+2)*(x-3)))
print("д) розклад на множники = ", factor(x^2 - 9*x + 20))
print("е) розв'язки рівняння = ", solve(x^2 - 4*x + 3 == 0, x))
```

### 4. Похідні, інтеграли та графіки функцій

а) побудуйте графік функції $f(x) = x^2 - 4x + 3$ на відрізку $[-5, 5]$; <br>
б) знайдіть похідну функції $f(x) = x^3 + 5x^2 - 2x + 7$; <br>
в) знайдіть невизначений інтеграл функції $2x^3 - x^2 + 5$; <br>
г) обчисліть визначений інтеграл $\int_{1}^{4} x^2\, dx$; <br>
д) оголосіть функцію $f(x) = 3x^2 - 2x + 1$ та обчисліть $f(5)$; <br>
е) побудуйте на одній площині графіки $f(x)=x^2$ і $g(x)=-x^2+5$ на $[-3,3]$.


```sage
var("x")
plot_a = plot(x^2 - 4*x + 3, (x, -5, 5))
print("а) Графік f(x) = x^2 - 4x + 3:")
show(plot_a)

print("б) похідна = ", diff(x^3 + 5*x^2 - 2*x + 7, x))

print("в) невизначений інтеграл = ", integrate(2*x^3 - x^2 + 5, x))

print("г) визначений інтеграл = ", integrate(x^2, x, 1, 4))

f(x) = 3*x^2 - 2*x + 1
print("д) f(5) = ", f(5))

plot_e = plot(x^2, (x, -3, 3), color='blue') + plot(-x^2 + 5, (x, -3, 3), color='red')
print("е) Графіки f(x)=x^2 та g(x)=-x^2+5:")
show(plot_e)
```

### 5. Програмування
а) напишіть умовний оператор, який визначає, чи число $n$ додатне, від'ємне або дорівнює нулю; <br>
б) напишіть цикл, що виводить числа від 1 до 10; <br>
в) напишіть функцію, яка перевіряє чи є задане число $n$ простим.

```sage
n = 7 

if n > 0:
    print("а) Число додатне")
elif n < 0:
    print("а) Число від'ємне")
else:
    print("а) Число дорівнює нулю")


print("б) Цикл:")
for i in range(1, 11):
    print(i)

def is_prime(number):
    if number <= 1:
        return False
    
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False 
            
    return True 

print("в) Перевірка функцією:")
print("Число 13 просте? ", is_prime(13))
print("Число 15 просте? ", is_prime(15))
```
