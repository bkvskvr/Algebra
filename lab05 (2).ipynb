### 1. Структура абелевих груп

Розглянемо абелеві групи порядку $400$:
$$
\mathbb{Z}_{400}, \quad \mathbb{Z}_{16}\times\mathbb{Z}_{25}, \quad \mathbb{Z}_{20}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{4}\times\mathbb{Z}_{100}, \quad \mathbb{Z}_{2}\times\mathbb{Z}_{10}\times\mathbb{Z}_{20}, \quad \mathbb{Z}_{505}^{*}.
$$
Для кожної з цих груп виконайте наступні кроки:
1. Знайдіть елементарні дільники групи.
2. Розкладіть групу у добуток циклічних $p$-груп.
3. Визначте кількість елементів кожного можливого порядку.
4. Порівняйте групи та визначте, які з них є ізоморфними.


```python
from itertools import product

groups = {
    'Z_400': [400],
    'Z_16 x Z_25': [16, 25],
    'Z_20 x Z_20': [20, 20],
    'Z_4 x Z_100': [4, 100],
    'Z_2 x Z_10 x Z_20': [2, 10, 20],
    'Z_505^*': [4, 100]
}

def elementary_divisors(cyclic_factors):
    ans = []
    for m in cyclic_factors:
        for p, e in factor(m):
            ans.append(ZZ(p)**ZZ(e))
    return sorted(ans)

def order_counts(cyclic_factors):
    exponent = lcm(cyclic_factors)

    at_most = {d: prod(gcd(d, m) for m in cyclic_factors)
               for d in divisors(exponent)}

    return {d: sum(moebius(d//e) * at_most[e] for e in divisors(d))
            for d in divisors(exponent)}

classes = {}
for name, factors in groups.items():
    ed = elementary_divisors(factors)
    counts = order_counts(factors)
    print('\n', name)
    print('Елементарні дільники:', ed)
    print('Розклад на циклічні p-групи:', ' x '.join('Z_%s' % x for x in ed))
    print('Кількість елементів кожного порядку:', counts)
    print('Перевірка порядку:', sum(counts.values()))
    classes.setdefault(tuple(ed), []).append(name)

print('\nКласи ізоморфізму:')
for names in classes.values():
    print(' ≅ '.join(names))
```

### 2. Одна абелева група порядку $n$
1. Реалізуйте функцію, яка для заданого натурального числа $n$ обчислює кількість попарно неізоморфних абелевих груп порядку $n$.
2. Знайдіть перші сто таких натуральних чисел $n$, для яких існує рівно одна абелева група порядку $n$. Для кожного знайденого $n$ проаналізуйте його розклад на прості множники.
3. Сформулюйте гіпотезу: для яких $n\in\mathbb{N}$ існує рівно одна (з точністю до ізоморфізму) абелева група порядку $n$.


```python
def number_of_abelian_groups(n):
    n = ZZ(n)
    if n < 1:
        raise ValueError('n має бути натуральним')
    return prod(Partitions(e).cardinality() for p, e in factor(n))

answer = []
n = 1
while len(answer) < 100:
    if number_of_abelian_groups(n) == 1:
        answer.append(n)
    n += 1

print('Перші 100 значень n:', answer)
print('Розклади на прості множники:')
for n in answer:
    print('%3s = %s' % (n, factor(n)))

print("Гіпотеза: n не ділиться на квадрат простого числа")

for n in answer:
    print(n, all(e == 1 for p, e in factor(n)))
```

### 3. Дільники нуля та одиниці в скінченних кільцях
Розгляньте скінченні кільця $\mathbb{Z}_{30}$, $\mathbb{Z}_{12}\times\mathbb{Z}_{15}$, $M_3(\mathbb{Z}_3)$ та для кожного з них виконайте наступні кроки:
1. Знайдіть дільники нуля та їх кількість.
2. Знайдіть дільники одиниці та їх кількість.
3. Порівняйте отримані результати та зробіть висновок про взаємозв'язок між дільниками нуля та дільниками одиниці в скінченних кільцях.


```python
from itertools import product

def inspect_commutative_ring(elements, zero, one, label):
    units = [a for a in elements if a.is_unit()]

    zero_divisors = [a for a in elements if a != zero and not a.is_unit()]
    print('\n' + label)
    print('Дільники одиниці:', units)
    print('Кількість дільників одиниці:', len(units))
    print('Ненульові дільники нуля:', zero_divisors)
    print('Кількість ненульових дільників нуля:', len(zero_divisors))
    print('Нуль окремо: разом з ним необоротних елементів', len(zero_divisors)+1)

A = Integers(30)
inspect_commutative_ring(list(A), A(0), A(1), 'Z_30')

B, C = Integers(12), Integers(15)
values = list(product(B, C))
units = [v for v in values if v[0].is_unit() and v[1].is_unit()]
zd = [v for v in values if v != (B(0), C(0)) and
      not (v[0].is_unit() and v[1].is_unit())]
print('\nZ_12 x Z_15')
print('Дільники одиниці:', units)
print('Кількість:', len(units))
print('Ненульові дільники нуля:', zd)
print('Кількість:', len(zd))

F = GF(3)
M = MatrixSpace(F, 3, 3)

mat_units, mat_zd = [], []
for entries in product(F, repeat=9):
    a = M(list(entries))
    if a.det() != 0:
        mat_units.append(a)
    elif a != M.zero():
        mat_zd.append(a)
print('\nM_3(Z_3)')
print('Кількість дільників одиниці:', len(mat_units))
print('Кількість ненульових дільників нуля:', len(mat_zd))
print('Приклади дільників одиниці:', mat_units[:3])
print('Приклади дільників нуля:', mat_zd[:3])
print('Висновок: у кожному скінченному кільці ненульові елементи')
print('поділяються на оборотні та дільники нуля; нуль необоротний.')
```

### 4. Факторкільця кільця многочленів
Нехай $f(x)=x^4+x+2$. Виконайте наступні кроки для кожного з кілець $R=\mathbb{Q}, \mathbb{Z}_4, \mathbb{Z}_5$:
1. Побудуйте ідеал $I=(f(x))$ у кільці $R[x]$.
2. Побудуйте факторкільце $R/I$ та знайдіть його порядок.
3. Визначте, чи є факторкільце $R/I$ полем.
4. Знайдіть дільники нуля у факторкільці $R/I$.
5. Розкладіь $f(x)$ на лінійні множники над $R$ та порівняйте з дільниками нуля у факторкільці $R/I$.
6. Перевірте, чи елемент $x+I\in R/I$ є оборотним.

```python
from itertools import product

for R, label in [(QQ, 'Q'), (Integers(4), 'Z_4'), (GF(5), 'Z_5')]:
    P = PolynomialRing(R, 'x')
    x = P.gen()
    f = x^4 + x + 2
    I = P.ideal(f)
    Q = P.quotient(I, 'a')
    a = Q.gen()

    print('\nR =', label)
    print('Ідеал:', I)
    print('Факторкільце:', Q)
    print('Порядок:', 'нескінченний' if R == QQ else R.cardinality()^4)
    print('Чи є полем:', R.is_field() and f.is_irreducible())
    print('Корені:', f.roots() if R.is_field() else [r for r in R if f(r) == 0])
    if R.is_field():
        print('Факторизація:', f.factor())
    else:
        print('Над Z_4 факторизація може бути нетривіальною')
    print('Чи оборотний x + I:', R(2).is_unit())

    if R == QQ and f.is_irreducible():
        print('Ненульових дільників нуля немає')
    else:
        zero_divisors = []
        for coeffs in product(R, repeat=4):
            z = Q(P(list(coeffs)))
            if z == 0:
                continue
            M = matrix(R, 4, 4, lambda i, j: (z*a^j).lift()[i])
            if not M.det().is_unit():
                zero_divisors.append(z)
        print('Кількість ненульових дільників нуля:', len(zero_divisors))
        print('Ненульові дільники нуля:', zero_divisors)
```
